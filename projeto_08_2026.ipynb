Identificamos que muitas colunas tem dados repetidos e não qualitativos, desnecessários para esta análise. Verificamos também que a chave em comum entre os arquivos é SQ_CANDIDATO, usaremos essa coluna para fazer a união dos arquivos.

# Preparando os Dados

```python
# Descobre quais colunas existem nos dois arquivos (exceto a coluna de união, ex: 'SQ_CANDIDATO')
colunas_repetidas = [col for col in con_cand.columns if col in bem_cand.columns and col != 'SQ_CANDIDATO']
```

```python
# Remove as colunas repetidas do segundo arquivo antes de juntar
con_cand_filtrado = con_cand.drop(columns=colunas_repetidas)
```

```python
# Une os arquivos usando a coluna 'SQ_CANDIDATO' como referência
# O 'how=left' garante que todos os dados do seu arquivo principal (bem_cand) sejam mantidos
resultado = pd.merge(bem_cand, con_cand_filtrado, on='SQ_CANDIDATO', how='left')
```

```python
# Mostra o resultado final com as colunas combinadas e sem repetições
resultado.head()
```

```python
resultado.info()
```

```python
# Mostra média, mínimo, máximo, desvio padrão e os quartis de todas as colunas
resultado.describe()
```

```python
# Agrupa por Partido e Sexo, e depois conta a quantidade de candidatos
subtotal_genero = resultado.groupby(['SG_PARTIDO', 'DS_GENERO']).size().reset_index(name='Total_Candidatos')

# Exibe a lista completa de contagem na tela
print(subtotal_genero)
```

```python
import matplotlib.pyplot as plt

# Conta a quantidade total de candidatos por sexo
contagem_genero = resultado['DS_GENERO'].value_counts()

# Configura o tamanho e gera o gráfico de pizza
plt.figure(figsize=(6, 6))
plt.pie(
    contagem_genero, 
    labels=contagem_genero.index, 
    autopct='%1.1f%%',     # Mostra a porcentagem com uma casa decimal
    startangle=90,         # Rotaciona o gráfico para começar no topo
    colors=['#3498db', '#e74c3c'] # Define cores bonitas (azul e vermelho)
)

# Adiciona um título ao gráfico
plt.title('Percentual de Candidatos por Gênero (Eleições 2026)', fontsize=14)

# Exibe o gráfico na tela do seu notebook
plt.show()
```
