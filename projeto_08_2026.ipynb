```python
import pandas as pd

# Carrega os dois arquivos .xls para a memória
bem_cand = pd.read_csv('bem_candidato_2026_SP.csv', sep=';', encoding='latin1')
con_cand = pd.read_csv('consulta_cand_2026_SP.csv', sep=';', encoding='latin1')
```

```python
# Descobre quais colunas existem nos dois arquivos (exceto a coluna de união, ex: 'SQ_CANDIDATO')
colunas_repetidas = [col for col in con_cand.columns if col in bem_cand.columns and col != 'SQ_CANDIDATO']
```

```python
# Remove as colunas repetidas do segundo arquivo antes de juntar
con_cand_filtrado = con_cand.drop(columns=colunas_repetidas)
```

```python
# Une os arquivos usando a coluna 'SQ_CANDIDATO' como referência
# O 'how=left' garante que todos os dados do seu arquivo principal (bem_cand) sejam mantidos
resultado = pd.merge(bem_cand, con_cand_filtrado, on='SQ_CANDIDATO', how='left')
```
