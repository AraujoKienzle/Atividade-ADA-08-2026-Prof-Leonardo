# Preparando os Dados

```python
# Identificar colunas com apenas 1 valor distinto (incluindo ou ignorando nulos)
colunas_constantes_cand = [col for col in con_cand.columns if con_cand[col].nunique() <= 1]
```

```python
display(colunas_constantes_cand)
```

```python
# Excluir colunas

base_limpa_cand = con_cand.drop(columns=colunas_constantes_cand)
base_limpa_cand.shape
```

Excluímos colunas irrelevantes do primeiro arquivo para esta análise. De 50 colunas agora o arquivo tem apenas 32 colunas. Permanece as 2.590 linhas.

```python
# Identificar colunas com apenas 1 valor distinto (incluindo ou ignorando nulos)
colunas_constantes_bem = [col for col in bem_cand.columns if bem_cand[col].nunique() <= 1]
```

```python
display(colunas_constantes_bem)
```

```python
base_limpa_bem = bem_cand.drop(columns=colunas_constantes_bem)
base_limpa_bem.shape
```

Excluímos colunas irrelevantes do segundo arquivo para esta análise. De 19 colunas agora o arquivo tem apenas 8 colunas. Permanece as 9.629 linhas.

Vamos fazer o cruzamento de dados (merge) cadastrais dos candidatos com a base dos bens declarados. Usaremos como chave primária a coluna SQ_CANDIDATO do arquivo base_limpa_cand e usar como chave secundária a coluna SQ_CANDIDATO do arquivo base_limpa_bem. 

```python
# Une os arquivos usando a coluna 'SQ_CANDIDATO' como referência
# O 'how=left' garante que todos os dados do seu arquivo principal (bem_cand) sejam mantidos
base_total = pd.merge(base_limpa_bem, base_limpa_cand, on='SQ_CANDIDATO', how='left')
```

```python
base_total.shape
```

```python
# Mostra o resultado final com as colunas combinadas e sem repetições
base_total.head()
```

```python
base_total.info()
```

```python
# Mostra média, mínimo, máximo, desvio padrão e os quartis de todas as colunas
base_total.describe()
```

```python
# Agrupa por Partido e Sexo, e depois conta a quantidade de candidatos
subtotal_genero = base_total.groupby(['SG_PARTIDO', 'DS_GENERO']).size().reset_index(name='Total_Candidatos')

# Exibe a lista completa de contagem na tela
print(subtotal_genero)
```

```python
import matplotlib.pyplot as plt

# Conta a quantidade total de candidatos por sexo
contagem_genero = base_total['DS_GENERO'].value_counts()

# Configura o tamanho e gera o gráfico de pizza
plt.figure(figsize=(6, 6))
plt.pie(
    contagem_genero, 
    labels=contagem_genero.index, 
    autopct='%1.1f%%',     # Mostra a porcentagem com uma casa decimal
    startangle=90,         # Rotaciona o gráfico para começar no topo
    colors=['#3498db', '#e74c3c'] # Define cores bonitas (azul e vermelho)
)

# Adiciona um título ao gráfico
plt.title('Percentual de Candidatos por Gênero (Eleições 2026)', fontsize=14)

# Exibe o gráfico na tela do seu notebook
plt.show()
```
