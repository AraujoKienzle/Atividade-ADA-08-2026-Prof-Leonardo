Vamos fazer o cruzamento de dados (merge) cadastrais dos candidatos com a base dos bens declarados. Usaremos como chave primária a coluna SQ_CANDIDATO do arquivo base_limpa_cand e usar como chave secundária a coluna SQ_CANDIDATO do arquivo base_limpa_bem. 

```python
# Une os arquivos usando a coluna 'SQ_CANDIDATO' como referência
# O 'how=left' garante que todos os dados do seu arquivo principal (bem_cand) sejam mantidos
base_total = pd.merge(base_limpa_bem, base_limpa_cand, on='SQ_CANDIDATO', how='left')
```

Ao tentar verificar o valor total de bens pro genero, notamos que a coluna VR_BEM_CANDIDATO estava como "str", o que impossibilita a soma, então seguimos os passos abaixo para trazer os valores do arquivo bem_cand (que eram numericos) com os dados da base_total.

```python
# Cria uma tabela contendo apenas as colunas necessárias de cada fonte
df_genero_chave = base_total[["SQ_CANDIDATO", "DS_GENERO"]].drop_duplicates()
df_bens_chave = bem_cand[["SQ_CANDIDATO", "VR_BEM_CANDIDATO"]].copy()
df_nome_chave = base_total[["SQ_CANDIDATO", "NM_CANDIDATO"]].drop_duplicates()
df_partido_chave = base_total[["SQ_CANDIDATO", "SG_PARTIDO"]].drop_duplicates()

# MERGE: Junta os bens ao gênero usando o SQ_CANDIDATO como chave
df_unificado = pd.merge(df_bens_chave, df_genero_chave, on="SQ_CANDIDATO", how="inner")
df_unificado = pd.merge(df_unificado, df_nome_chave, on="SQ_CANDIDATO", how="inner")
df_unificado = pd.merge(df_unificado, df_partido_chave, on="SQ_CANDIDATO", how="inner")


# LIMPEZA DOS VALORES: Converte o texto com vírgula para número decimal
df_unificado["VR_BEM_NUMERICO"] = df_unificado["VR_BEM_CANDIDATO"].astype(str).str.strip()
df_unificado["VR_BEM_NUMERICO"] = (
    df_unificado["VR_BEM_NUMERICO"]
    .str.replace(".", "", regex=False)   # Remove pontos de milhar se houver
    .str.replace(",", ".", regex=False)   # Troca a vírgula pelo ponto decimal do Python
)

# Converte de texto para formato numérico de verdade
df_unificado["VR_BEM_NUMERICO"] = pd.to_numeric(df_unificado["VR_BEM_NUMERICO"], errors="coerce")
df_unificado["VR_BEM_NUMERICO"] = df_unificado["VR_BEM_NUMERICO"].fillna(0)

# CALCULO DA SOMA TOTAL POR GÊNERO
soma_por_genero = df_unificado.groupby("DS_GENERO")["VR_BEM_NUMERICO"].sum()

# Imprime o resultado final limpo e formatado em Reais (R$)
print("=== VALOR TOTAL DE BENS POR GÊNERO (CRUZAMENTO POR CHAVE) ===")
for genero, valor in soma_por_genero.items():
    valor_ptbr = f"R$ {valor:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")
    print(f"{genero}: {valor_ptbr}")
```

```python
display(df_unificado)
```

```python
# Agrupa os dados do df_unificado por candidato (SQ_CANDIDATO)
# Isso consolida toda a riqueza da mesma pessoa e preserva o Nome e o Partido
ranking_dados = (
    df_unificado.groupby("SQ_CANDIDATO")
    .agg(
        {
            "VR_BEM_NUMERICO": "sum",  # Soma todos os bens individuais do candidato
            "NM_CANDIDATO": "first",  # Pega o nome dele
            "SG_PARTIDO": "first",  # Pega a sigla do partido dele
        }
    )
    .reset_index()
)

# Ordena o patrimônio total do maior para o menor e isola os 10 primeiros
top_10_ricos = ranking_dados.sort_values(
    by="VR_BEM_NUMERICO", ascending=False
).head(10)

# 3. Imprime o resultado final formatado no padrão brasileiro de moeda (R$)
print("=== TOP 10 CANDIDATOS MAIS RICOS ===")
for i, (_, linha) in enumerate(top_10_ricos.iterrows(), 1):
    valor_ptbr = (
        f"R$ {linha['VR_BEM_NUMERICO']:,.2f}"
        .replace(",", "X")
        .replace(".", ",")
        .replace("X", ".")
    )
    print(
        f"{i}º. {linha['NM_CANDIDATO']} ({linha['SG_PARTIDO']}): {valor_ptbr}"
    )
```

```python
analise_idade = (
    base_total
    .dropna(subset=['FAIXA_ETARIA'])
    .groupby('FAIXA_ETARIA', observed=True)['VR_BEM_CANDIDATO']
    .agg(['count', 'sum', 'mean', 'median'])
    .sort_values('median', ascending=False)
)

analise_idade
```
